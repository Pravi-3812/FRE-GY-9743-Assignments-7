### Include an IFR curve plot and briefly explain how your method ensures monotonicity and convexity.

```python
import numpy as np
import matplotlib.pyplot as plt

def make_model(method):
    bms = qfCreateModelBuildMethodCollection([
        qfCreateBuildMethod(kind, {**inputs, "INTERPOLATION METHOD": method, "EXTRAPOLATION METHOD": "FLAT"})
        for kind, inputs in curve_inputs])
    return qfCreateModel("2026-01-02", "YIELD_CURVE", data_collection, bms)

pcp_model = make_model("PIECEWISE_CONSTANT_LEFT_CONTINUOUS")
mc_model = make_model("MONOTONE_CONVEX")

ts = np.linspace(0, 5.5, 551)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, name in zip(axes, ["SOFR-1B", "FF-1B", "SOFR-1B-FLAT"]):
    pcp = pcp_model.retrieve_model_component(name).state_data_interpolator
    mc = mc_model.retrieve_model_component(name).state_data_interpolator
    ax.plot(ts, [100 * pcp.interpolate(t) for t in ts], label="Piecewise constant")
    ax.plot(ts, [100 * mc.interpolate(t) for t in ts], label="Monotone convex")
    ax.scatter(mc.axis1, 100 * np.asarray(mc.values), color="black", zorder=3, label="Nodes")
    ax.set_title(name); ax.set_xlabel("Years"); ax.set_ylabel("IFR (%)")
axes[0].legend()
plt.tight_layout()
plt.show()
```

**Method.** Inspired by Hagan and West's basic quadratic interpolator, I use **one quadratic on each interval** $[t_i, t_{i+1}]$:

$$f(t) = f_i + d\,y + (s_i - d)\,\frac{y^2}{h_i}, \qquad y = t - t_i,\; h_i = t_{i+1}-t_i,\; s_i = \frac{f_{i+1}-f_i}{h_i}.$$

* **Passes through the points** for any starting slope $d$: $f(t_i)=f_i$ and $f(t_{i+1}) = f_{i+1}$.
* The slope goes in a straight line from $d$ to $2s_i - d$, so:
  * **convex** ($f'' = 2(s_i-d)/h_i \ge 0$) when $d \le s_i$;
  * **monotone** (the slope keeps the sign of $s_i$) when $d$ lies between $0$ and $2s_i$.
* **Choice of $d$:** each quadratic starts with the slope the previous one ended with, so the curve is smooth at the nodes.
  The first interval starts with $d = s_0$ (a straight line). The slope is then **clipped** to $[\min(0, 2s_i),\, s_i]$,
  which guarantees monotone and convex on every interval.
* **Integral:** a quadratic integrates to a cubic, so `integrate` is an exact formula.

**Data that cannot satisfy everything.** When the carried-over slope is outside the allowed range (a hump, a flattening
curve, or a sharp change in slope), clipping keeps the fit, continuity and the monotone/convex shape on the interval, at the
cost of a small kink at that node. A single node gives a constant (the funding curve), and extrapolation is flat.
On the supplied data the SOFR curve is smooth and convex everywhere; the FF curve needs a clip at two nodes, leaving tiny
kinks (about 0.5 and 4.6 bp per year in the slope) that are barely visible in the plot.

**PV comparison.** On these upward-sloping curves the left-continuous step function uses the *next* (higher) node value on
each interval, while the monotone-convex curve rises gradually between nodes. So its forward rates are lower between nodes.

the steep 2028–2029 part of the curve. In the basis swap both legs fall, but SOFR falls slightly more than FF (its curve is
steeper), so receiving SOFR + 10bp loses a little value.
